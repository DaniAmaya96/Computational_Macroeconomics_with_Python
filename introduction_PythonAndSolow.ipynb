# 3. From the Solow Model to Python

Now that we know the basic Python tools that we need, we can use them to simulate our first macroeconomic model.

You already know the theory of the Solow growth model. Therefore, our objective here is not to derive the model again, but to understand how we can translate its equations into **computational instructions**.

For simplicity, suppose that there is no population growth and no technological growth.

The production function per worker is:

$$
y_t = A k_t^\alpha
$$

A constant fraction $s$ of output is saved:

$$
i_t = s y_t
$$

and capital depreciates at rate $\delta$.

Therefore, the law of motion of capital is:

$$
k_{t+1}
=
s A k_t^\alpha
+
(1-\delta)k_t
$$

This equation is particularly useful computationally because it tells us exactly how to obtain tomorrow's capital once we know today's capital.

In other words:

$$
k_t
\longrightarrow
k_{t+1}
$$

This is precisely the type of operation that a computer can repeat using a loop.

## 3.1 Calibrating the Model

Before solving the model, we need to assign numerical values to its parameters.

In computational economics, this step is usually called **calibration**.

````python
A = 1.0
alpha = 0.33
s = 0.25
delta = 0.10

k0 = 0.5
T = 250
````

Here:

* `A` represents productivity,
* `alpha` represents the capital share,
* `s` represents the saving rate,
* `delta` represents the depreciation rate,
* `k0` represents the initial level of capital,
* and `T` is the number of periods that we want to simulate.

Notice something important, we do not need a discount factor $\beta$ in the Solow model.

Why?

Because households are not solving an intertemporal optimization problem. The saving rate $s$ is given exogenously.

This will change when we move from the Solow model to the Ramsey model, you will see it with Andre, don't worry.

## 3.2 Defining the Production Function

We already learned how to define functions in Python.

The production function:

$$
y = Ak^\alpha
$$

can be written as:

````python
def production(k, A, alpha):

    y = A * k**alpha

    return y
````

We can test the function for a particular level of capital.

````python
production(10, A, alpha)
````

The arguments of the function have a direct economic interpretation:

* `k` is capital per worker,
* `A` is productivity,
* `alpha` is the capital share.

An important advantage of writing the production function in this way is that we can use the same function many times without rewriting the equation.

Later, we will also see that the same function can receive either a single number or an entire NumPy array.

# 4. Simulating the Solow Model

The law of motion of capital is:

$$
k_{t+1}
=
sA k_t^\alpha
+
(1-\delta)k_t
$$

Suppose initially that:

$$
k_0 = 0.5
$$

We first create a list containing our initial capital.

````python
k_path = [k0]
````

At this moment the list contains only:

$$
[k_0]
$$

Therefore, `k_path[0]` represents:

$$
k_0
$$

Now we want Python to calculate:

$$
k_1,k_2,k_3,\ldots,k_T
$$

using the same law of motion repeatedly.

So we are going to use a  `for` loop 

````python
k_path = [k0]

for t in range(T):

    k_next = (
        s * production(k_path[t], A, alpha)
        + (1 - delta) * k_path[t]
    )

    k_path.append(k_next)


len(k_path)
````

Let's understand the algorithm.

In the first iteration:

```python
t = 0
```

Therefore, Python reads:

```python
k_path[t]
```

as:

```python
k_path[0]
```

which is our initial capital $k_0$.

Python then calculates:

$$
k_1
=
sAk_0^\alpha
+
(1-\delta)k_0
$$

The instruction:

```python
k_path.append(k_next)
```

adds the newly calculated capital to the list.

Therefore, after the first iteration, our list contains:

$$
[k_0,k_1]
$$

In the second iteration:

```python
t = 1
```

and Python uses `k_path[1]`, which is $k_1$, to calculate:

$$
k_2
=
sAk_1^\alpha
+
(1-\delta)k_1
$$

The computer repeats exactly the same operation until we obtain:

$$
k_0,k_1,k_2,\ldots,k_T
$$

Therefore, a `for` loop can be interpreted economically as the repeated application of a **difference equation**.

````python
print("Initial capital:", k_path[0])
print("Capital after 1 period:", k_path[1])
print("Capital after 250 periods:", k_path[-1])
````

# 6. The Steady State

You already know that the steady state is a situation in which capital per worker does not change.

Therefore:

$$
k_{t+1}=k_t=k^*
$$

Using the Solow law of motion:

$$
k^*
=
sA(k^*)^\alpha
+
(1-\delta)k^*
$$

Subtracting $(1-\delta)k^*$ from both sides:

$$
k^*-(1-\delta)k^*
=
sA(k^*)^\alpha
$$

Therefore:

$$
\delta k^*
=
sA(k^*)^\alpha
$$

Dividing by $(k^*)^\alpha$:

$$
\delta(k^*)^{1-\alpha}
=
sA
$$

Therefore:

$$
(k^*)^{1-\alpha}
=
\frac{sA}{\delta}
$$

Finally:

$$
\boxed{
k^*
=
\left(
\frac{sA}{\delta}
\right)^{\frac{1}{1-\alpha}}
}
$$

Because this model has an analytical solution, we can calculate the steady state directly in Python.

````python
s= 0.25

k_ss = (s * A / delta)**(1 / (1 - alpha))

print(k_ss)
````

### Visualizing the Steady State

We now have a number, but a number on its own does not teach very much. Let us go back to the trajectory that we simulated in Section 4 and draw $k^*$ on top of it.

A new instruction: `plt.axhline(value)` draws a **horizontal line** across the whole figure. It is the natural way to mark a constant, such as a steady state.

````python
plt.figure(figsize=(9, 5))

plt.plot(k_path, linewidth=2, label="Capital path $k_t$")
plt.axhline(k_ss, color="red", linestyle="--", label="Steady state $k^*$")

plt.xlabel("Time")
plt.ylabel("Capital per worker")
plt.title("The economy converges to its steady state")
plt.legend()
plt.grid(alpha=0.3)

plt.show()
````

This is convergence, seen directly.

The capital path rises quickly at first and then flattens onto the dashed line. It approaches $k^*$ but never crosses it. The reason is economic, not numerical: the production function is concave, so when capital is scarce its marginal product is high and each unit of investment adds a lot. As capital accumulates, the marginal product falls, and eventually the investment that the economy generates is only just enough to replace what depreciates. At that point capital stops changing.

Notice one subtle thing. After 250 periods the capital path matches $k^*$ to six decimal places, but it never *exactly* equals it: each period the economy closes a fixed fraction of the remaining distance, so the gap shrinks toward zero without ever becoming zero. Convergence here is **asymptotic**.

That is precisely why, in the next section, we will not be able to ask the computer whether $k_{t+1}$ equals $k_t$. We will have to ask whether they are close enough — which is the tolerance we introduced in Section 2.10, now doing real work.

## 6.1 Finding the Steady State Numerically

Until now, we have obtained the steady state using its analytical solution:

$$
k^*
=
\left(
\frac{sA}{\delta}
\right)^{\frac{1}{1-\alpha}}
$$

However, many macroeconomic models do not have a closed-form solution.

Therefore, we also need to learn how to find a steady state **numerically**.

Remember that the steady state satisfies:

$$
k_{t+1}=k_t
$$

Computationally, because of numerical precision, we will instead look for a value such that:

$$
|k_{t+1}-k_t|<\varepsilon
$$

where $\varepsilon$ is our tolerance.

Instead of deciding in advance how many periods to simulate, we can tell Python:

> Continue iterating while capital is still changing by more than our tolerance.

For this purpose, we use a `while` loop.

````python
k_old = 0.5

tol = 1e-8
error = 1.0

iteration = 0
max_iter = 10000

while error > tol and iteration < max_iter:

    k_new = (
        s * production(k_old, A, alpha)
        + (1 - delta) * k_old
    )

    error = abs(k_new - k_old)

    k_old = k_new

    iteration = iteration + 1
````

Let's read the algorithm step by step.

First, we start from an initial guess:

$$
k^{(0)}=0.5
$$

At every iteration, Python applies the Solow law of motion:

$$
k^{(n+1)}
=
sA(k^{(n)})^\alpha
+
(1-\delta)k^{(n)}
$$

Then, we calculate the difference between the new and old values:

$$
\text{error}
=
|k^{(n+1)}-k^{(n)}|
$$

If:

$$
\text{error}>\varepsilon
$$

Python continues iterating.

If:

$$
\text{error}<\varepsilon
$$

we consider that the algorithm has converged.

Notice the difference between `for` and `while`:

- `for` repeats an operation a predetermined number of times.
- `while` repeats an operation until a condition is satisfied.

````python
print("Numerical steady state:", k_old)
print("Analytical steady state:", k_ss)
print("Number of iterations:", iteration)
print("Final convergence error:", error)
print("Difference:", abs(k_old - k_ss))
````

We should obtain:

$$
k^*_{\text{numerical}}
\approx
k^*_{\text{analytical}}
$$

This illustrates one of the most important principles of computational economics:

> **Whenever we have an analytical solution, we should use it to validate our numerical algorithm.**

The computer can produce a number even when our code is wrong. Therefore, numerical results should be tested whenever possible against a case in which we already know the correct answer.

The structure of our algorithm is:

$$
\text{initial guess}
\rightarrow
\text{apply the law of motion}
\rightarrow
\text{calculate the error}
\rightarrow
\text{update}
\rightarrow
\text{repeat}
$$

This is our first example of a **fixed-point algorithm**. Don't worry if you don't understand it, later I will explain it to you. 

In the Solow model, we iterate a number:

$$
k
$$

Later in the course, when we study dynamic programming, we will use a similar logic to iterate a function:

$$
V(k)
$$

# 7. The Solow Diagram

We have now found the steady state three times: with a formula, with an algorithm, and as a horizontal line that the economy approaches. But we still have not seen **why** it sits exactly where it sits.

The Solow diagram answers that question, and it does so by changing what we put on the horizontal axis. Until now the horizontal axis was **time**. Now it will be **capital**.

The trick is to rewrite the law of motion. Starting from

$$
k_{t+1} = sAk_t^{\alpha} + (1-\delta)k_t
$$

and subtracting $k_t$ from both sides:

$$
\underbrace{k_{t+1}-k_t}_{\text{change in capital}}
=
\underbrace{sAk_t^{\alpha}}_{\text{what we add}}
-
\underbrace{\delta k_t}_{\text{what we lose}}
$$

Read that equation carefully, because the whole model is in it:

* Capital **grows** whenever investment $sAk^{\alpha}$ exceeds depreciation $\delta k$.
* Capital **falls** whenever depreciation exceeds investment.
* Capital is **constant** exactly where the two are equal.

So if we draw both objects as functions of $k$, the steady state is simply the point where the curve crosses the line.

````python
# A grid of capital values on which to evaluate the two curves
k_grid = np.linspace(0.01, 10, 500)

output       = production(k_grid, A, alpha)   # f(k)
investment   = s * output                     # s f(k)
depreciation = delta * k_grid                 # delta k

plt.figure(figsize=(9, 6))

plt.plot(k_grid, output,       linewidth=2, label=r"Output  $f(k) = Ak^{\alpha}$")
plt.plot(k_grid, investment,   linewidth=2, label=r"Investment  $s\,f(k)$")
plt.plot(k_grid, depreciation, linewidth=2, label=r"Depreciation  $\delta k$")

plt.axvline(k_ss, color="gray", linestyle="--", label=r"Steady state  $k^*$")

plt.xlabel("Capital per worker $k$")
plt.ylabel("Units of output per worker")
plt.title("The Solow Diagram")
plt.legend()
plt.grid(alpha=0.3)

plt.show()
````

Three things are worth reading off this figure.

**First, the shapes are economics, not decoration.** Output $f(k)$ is concave because capital has diminishing returns: the first machine transforms a workshop, the hundredth barely helps. Investment $s\,f(k)$ is that same curve scaled down by the saving rate, so it is concave too. Depreciation $\delta k$, on the other hand, is a **straight line**: we always lose the same fraction of whatever stock we have. A concave curve that starts steep and a straight line that starts flat must cross exactly once. That single crossing is why the steady state exists and why it is unique.

**Second, the vertical distance between $f(k)$ and $s\,f(k)$ is consumption.** Everything that is not saved is consumed. Later in the course, when households actually get to choose, this distance stops being a fixed proportion and becomes the object of a decision.

**Third, the dashed line falls exactly on the intersection.** That is the same $k^*$ we obtained from the formula and from the `while` loop. Four different routes to the same number.

### Why does the economy move toward $k^*$?

The diagram does not only tell us where the steady state is. It tells us in which direction the economy travels from anywhere else.

````python
k_grid = np.linspace(0.01, 10, 500)

plt.figure(figsize=(9, 6))

plt.plot(k_grid, s * production(k_grid, A, alpha), linewidth=2, label=r"Investment  $s\,f(k)$")
plt.plot(k_grid, delta * k_grid, linewidth=2, label=r"Depreciation  $\delta k$")
plt.axvline(k_ss, color="gray", linestyle="--", label=r"Steady state  $k^*$")

# plt.annotate writes text on the figure; arrowprops turns it into an arrow
plt.annotate("", xy=(3.5, 0.10), xytext=(1.2, 0.10),
             arrowprops=dict(arrowstyle="->", linewidth=2, color="darkgreen"))
plt.text(1.0, 0.17, "investment > depreciation\ncapital grows", color="darkgreen")

plt.annotate("", xy=(4.4, 0.10), xytext=(8.5, 0.10),
             arrowprops=dict(arrowstyle="->", linewidth=2, color="firebrick"))
plt.text(5.6, 0.17, "depreciation > investment\ncapital falls", color="firebrick")

plt.xlabel("Capital per worker $k$")
plt.ylabel("Units of output per worker")
plt.title("Why the economy converges: the direction of motion")
plt.legend(loc="upper left")
plt.grid(alpha=0.3)

plt.show()
````

To the **left** of $k^*$ the investment curve lies above the depreciation line, so the economy adds more capital than it loses and moves to the right. To the **right** of $k^*$ the depreciation line is above the investment curve, so the economy loses more than it adds and moves to the left.

Both arrows point toward the same place. This is what we mean when we say the steady state is **globally stable**: the economy converges to $k^*$ from any starting point. It is also, incidentally, the reason our `while` loop worked no matter what initial guess we gave it. Try changing `k_old = 0.5` to `k_old = 30` in Section 6.1 and running it again: it lands on exactly the same number.

### The diagram explains our earlier activity

Remember the exercise where we raised the saving rate from $s = 0.25$ to $s = 0.35$ and the capital path shifted upward. Now we can see exactly why.

````python
k_grid = np.linspace(0.01, 12, 500)

k_ss_1 = (s1 * A / delta)**(1 / (1 - alpha))
k_ss_2 = (s2 * A / delta)**(1 / (1 - alpha))

plt.figure(figsize=(9, 6))

plt.plot(k_grid, s1 * production(k_grid, A, alpha), linewidth=2, label=r"Investment with $s = 0.25$")
plt.plot(k_grid, s2 * production(k_grid, A, alpha), linewidth=2, label=r"Investment with $s = 0.35$")
plt.plot(k_grid, delta * k_grid, linewidth=2, color="black", label=r"Depreciation  $\delta k$")

plt.axvline(k_ss_1, color="gray", linestyle="--")
plt.axvline(k_ss_2, color="gray", linestyle="--")

plt.xlabel("Capital per worker $k$")
plt.ylabel("Units of output per worker")
plt.title("A higher saving rate shifts the investment curve up and moves $k^*$ to the right")
plt.legend(loc="upper left")
plt.grid(alpha=0.3)

plt.show()

y_ss_1 = production(k_ss_1, A, alpha)
y_ss_2 = production(k_ss_2, A, alpha)

print(f"With s = {s1}:  k* = {k_ss_1:.3f}   y* = {y_ss_1:.3f}")
print(f"With s = {s2}:  k* = {k_ss_2:.3f}   y* = {y_ss_2:.3f}")
print()
print(f"Capital rises by {100*(k_ss_2/k_ss_1 - 1):.1f}%")
print(f"Output  rises by {100*(y_ss_2/y_ss_1 - 1):.1f}%")
````

A higher saving rate does not change depreciation, and it does not change the production function. It simply lifts the investment curve, so the crossing point slides to the right. That is the entire content of the comparative-statics exercise we ran earlier, in one picture.

But now look at the numbers printed under the figure, because they contain the single most important lesson of this model.

Raising the saving rate by 40% (from 0.25 to 0.35) raises steady-state **capital** by about 65%. That sounds like a lot. But steady-state **output** rises by only about 18%, and output is what people actually consume. The reason is the concavity we have been talking about all along: that extra capital is being added precisely where the production function has already flattened out, so each additional machine buys very little additional output.

This is why the Solow model, taken literally, has a hard time explaining why some countries are rich and others are poor. Saving more helps, but it runs into diminishing returns very quickly — a point we will confront directly with real data in this week's activity.

### Where this is going

Notice what the investment curve $s\,f(k)$ really is: it is a **decision rule**. It says how much this economy sets aside for tomorrow, for every level of capital it might have today. In the Solow model we simply assumed that rule was a constant fraction of output, and we drew it.

In the models that come next, nobody hands us that rule. Households choose it, weighing consumption today against consumption tomorrow, and the decision at every level of capital depends on what they plan to do at every other level. Our job will be to **find** that curve rather than assume it.

That is what the Bellman equation is for. And when we go looking for it, we will use the very same `while` loop we wrote in Section 6.1 — the difference being that what stops changing will no longer be a number, but an entire function.

# 8. The problem of the day: a fixed payment abroad

In the slides we solved this problem on the board: every period the country pays a fixed amount $p > 0$ per worker to a foreign country, in exchange for protection. Output is used for consumption, investment or the payment, so investment becomes $s\,y_t - p$ and the law of motion of capital is

$$
k_{t+1} = sAk_t^{\alpha} - p + (1-\delta)k_t
$$

It is the same law of motion as in Section 4, with **one term more**. We found that the economy can have two steady states, $k_L$ and $k_H$, and that $k_L$ is a threshold: an economy that starts below it collapses. On the board we could not compute $k_L$ and $k_H$, because there is no formula for them. Now we use the computer, with the tools of this session.

We keep the calibration of the whole session and set the payment to $p = 0.12$.

````python
s = 0.25
p = 0.12

print(f"A = {A}, alpha = {alpha}, s = {s}, delta = {delta}, p = {p}")
````

## 8.1 The Solow diagram with the payment

We draw the diagram of Section 7 again, changing one line: the investment curve is now $s f(k) - p$. Subtracting a constant moves the whole curve **down** by $p$.

````python
k_grid = np.linspace(0.01, 3, 500)

investment_with_p = s * production(k_grid, A, alpha) - p
depreciation = delta * k_grid

plt.figure(figsize=(6, 7))
plt.plot(k_grid, s * production(k_grid, A, alpha), linewidth=2, linestyle="--", color="gray", label=r"Investment without the payment  $s\,f(k)$")
plt.plot(k_grid, investment_with_p, linewidth=2, label=r"Investment with the payment  $s\,f(k) - p$")
plt.plot(k_grid, depreciation, linewidth=2, label=r"Depreciation  $\delta k$")
plt.axhline(0, color="black", linewidth=0.8)
plt.xlabel("Capital per worker $k$")
plt.ylabel("Units of output per worker")
plt.legend(loc="upper left")
plt.grid(alpha=0.3)
plt.show()
````

Now the investment curve starts **below** the depreciation line. It crosses the line twice: first at a small level of capital, $k_L$, and then again at $k_H$. In Section 7 we argued that a concave curve that starts above a straight line crosses it exactly once. The payment breaks that argument, because the curve no longer starts above the line.

Read the directions as in Section 7:

- Between $k_L$ and $k_H$, investment is above depreciation: capital grows.
- To the right of $k_H$, depreciation is above investment: capital falls. So $k_H$ attracts.
- To the left of $k_L$, depreciation is above investment: capital falls, away from $k_L$. So $k_L$ repels.

## 8.2 The `while` loop finds $k_H$, but not $k_L$

Let's use the fixed-point algorithm of Section 6.1, with the new law of motion. We add one safeguard with an `if`: if capital becomes negative, the economy has collapsed and we stop iterating.

````python
tol = 1e-8
max_iter = 10000

def iterate_to_steady_state(k_start):
    """Fixed-point iteration of Section 6.1, with the payment p."""
    k_old = k_start
    error = 1.0
    iteration = 0
    while error > tol and iteration < max_iter:
        k_new = s * production(k_old, A, alpha) - p + (1 - delta) * k_old
        if k_new <= 0:
            return 0.0, iteration          # capital exhausted
        error = abs(k_new - k_old)
        k_old = k_new
        iteration = iteration + 1
    return k_old, iteration

for k_start in [0.5, 1.0, 3.0, 0.10]:
    k_final, n_iter = iterate_to_steady_state(k_start)
    print(f"Starting at k = {k_start:4.2f}  ->  k = {k_final:.6f} after {n_iter} iterations")
````

From 0.5, 1.0 or 3.0 the loop always lands on the same number: that is $k_H$. From 0.10 the economy collapses to zero. No starting point ever lands on $k_L$.

The reason is the slope of the law of motion at each steady state, $\phi'(k) = sA\alpha k^{\alpha-1} + 1 - \delta$. At $k_H$ the slope is smaller than one, so each iteration closes part of the gap, as in Section 6. At $k_L$ the slope is **larger than one**, so each iteration multiplies any small gap instead of closing it. An iteration can only find a steady state that attracts.

## 8.3 Finding $k_L$ on a grid

To find $k_L$ we use a different idea. Both steady states satisfy

$$
s A k^{\alpha} - p - \delta k = 0
$$

so we evaluate the left-hand side on a fine grid of capital values and look for the place where it changes sign, from negative to positive. We only need a `for` loop and an `if`, two tools we already know.

````python
k_grid = np.linspace(0.01, 1.0, 10001)
gap = s * production(k_grid, A, alpha) - p - delta * k_grid

for i in range(len(k_grid) - 1):
    if gap[i] < 0 and gap[i + 1] > 0:
        k_L = k_grid[i]

k_H, _ = iterate_to_steady_state(1.0)

slope_L = s * A * alpha * k_L**(alpha - 1) + 1 - delta
slope_H = s * A * alpha * k_H**(alpha - 1) + 1 - delta

print(f"k_L is approximately {k_L:.4f}   (slope of the law of motion: {slope_L:.3f})")
print(f"k_H                 = {k_H:.4f}   (slope of the law of motion: {slope_H:.3f})")
````

The grid gives $k_L$ only up to its spacing, $10^{-4}$. That is enough for today. In the next session we will learn **bisection**, a method that finds a sign change to any precision we want, and we will come back to this $k_L$.

The slopes confirm the argument of Section 8.2: about 1.19 at $k_L$ and 0.95 at $k_H$.

## 8.4 The poverty trap

Finally, we simulate two economies with the `for` loop of Section 4. One starts 10% above $k_L$ and the other 10% below. If capital becomes negative we set it to zero: the economy has nothing left.

````python
T = 250
starts = [1.10 * k_L, 0.90 * k_L]
paths = []

for k0_trap in starts:
    k_path = [k0_trap]
    for t in range(T):
        k_next = s * production(k_path[t], A, alpha) - p + (1 - delta) * k_path[t]
        if k_next < 0:
            k_next = 0.0
        k_path.append(k_next)
    paths.append(k_path)

plt.figure(figsize=(9, 5))
plt.plot(paths[0], linewidth=2, label="Starts 10% above $k_L$")
plt.plot(paths[1], linewidth=2, label="Starts 10% below $k_L$")
plt.axhline(k_H, color="gray", linestyle="--", label="$k_H$")
plt.axhline(k_L, color="gray", linestyle=":", label="$k_L$")
plt.xlabel("Time")
plt.ylabel("Capital per worker")
plt.legend(loc="center right")
plt.grid(alpha=0.3)
plt.show()
````

Two economies that start almost at the same point, one with 18% less capital than the other, end up in completely different places. The one above $k_L$ converges to $k_H$, getting within 1% of it after about 130 periods; the one below cannot pay $p$ with its savings, eats its capital and reaches zero in 12 periods.

This is a **poverty trap**: history, the initial capital stock, decides the destiny of the economy. It is also a good reminder of what we learned today about computation: a steady state that repels exists, but no iteration will ever find it for us.
